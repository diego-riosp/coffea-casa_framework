```python
import awkward as ak
import numpy as np
from coffea.analysis_tools import Weights, PackedSelection
import correctionlib
import hist
from coffea import processor
from coffea.nanoevents import NanoAODSchema, NanoEventsFactory
from constructors.filesets import filesets
from loadmodule import *
from coffea.lookup_tools import txt_converters, rochester_lookup
from coffea.lumi_tools import LumiMask
from coffea.nanoevents.methods import candidate
import warnings
warnings.filterwarnings("ignore", message="Missing cross-reference index ")
test_fileset = getattr(filesets,'test_fileset')
events = NanoEventsFactory.from_root(
    test_fileset["DoubleMuon"]["files"],
    schemaclass=NanoAODSchema,
    entry_stop=10_000,
).events();
```

```python
import yaml
loadAll("constructors/utils/util_functions.py")

class SelectionEngine:
    def __init__(self, cfg, context):
        self.cfg = cfg
        self.context = context   # variables disponibles (events, dileptons, etc)
        self.objects = {}

    def eval_expr(self, expr):
        return eval(expr, {}, self.context | {"objects": self.objects})

    def apply(self):
        for name, block in self.cfg["objects"].items():
            # construir objeto
            obj_expr = block["object"]
            obj = self.eval_expr(obj_expr)

            # aplicar cortes
            mask = None
            for cut in block["cuts"]:
                m = self.eval_expr(cut)
                mask = m if mask is None else (mask & m)

            obj = obj[mask]

            # guardar
            self.objects[name] = obj

        return self.objects
```

```python
loadAll("constructors/selections/object_selector.py")
muons = events.Muon
muons_mask = (
    (muons.pt > 10)
    & (abs(muons.eta) < 2.4)
    & muons.tightId
    & (muons.pfRelIso03_all < 0.15)
)
dimuons = dileptons(muons)
diobjects_mask = (
    (dimuons.l1.delta_r(dimuons.l2) > 0.3)
    & (dimuons.l1.charge * dimuons.l2.charge < 0)
    & (dimuons.p4.mass > 60.0) & (dimuons.p4.mass < 120.0)
)
print(ak.sum(ak.num(ObjectSelector(events).objectSelector(muons, muons_mask))))
print(ak.sum(ak.num(ObjectSelector(events).objectSelector(dimuons, diobjects_mask))))
```

```python
################## object correction
#apply_rochester_corrections_run2(events,"2018",rochester)
################## object selection
is_mc = "genWeight" in events.fields
muons = events.Muon

good_muons = (
    (muons.pt > 10)
    & (abs(muons.eta) < 2.4)
    & muons.tightId
    & (muons.pfRelIso03_all < 0.15)
)
muons = muons[good_muons]

lorentz_muons = ak.zip(
    {
        "pt": muons.pt,
        "eta": muons.eta,
        "phi": muons.phi,
        "mass": muons.mass,
        "charge": muons.charge,
    },
    with_name="PtEtaPhiMCandidate",
    behavior=candidate.behavior,
)
dimuons = ak.combinations(lorentz_muons, 2, fields=["lead", "trail"])
dimuons = dimuons[ak.argsort(dimuons.lead.pt, axis=1)]
dimuons["p4"] = dimuons.lead + dimuons.trail
dimuons["pt"] = dimuons.p4.pt

good_dimuons = (
    (dimuons.lead.delta_r(dimuons.trail) > 0.3)
    & (dimuons.lead.charge * dimuons.trail.charge < 0)
    & (dimuons.p4.mass > 60.0) & (dimuons.p4.mass < 120.0)
)
dimuons = dimuons[good_dimuons]

################## event selection
#sumw_before = ak.sum(events.genWeight)
if is_mc:
    sumw_before = ak.sum(events.genWeight)
else:
    sumw_before = len(events)
    
selections = PackedSelection()
if is_mc:
    lumi_mask = np.ones(len(events), dtype="bool")
else:
    lumi_info = LumiMask("constructors/selections/sets/Cert_Collisions2022_355100_362760_Golden.txt")
    lumi_mask = lumi_info(events.run, events.luminosityBlock) == 1
selections.add("golden_json", lumi_mask)
selections.add("npvsGood", events.PV.npvsGood>0)
selections.add("IsoMu24", events.HLT.IsoMu24)
selections.add("two_muons", ak.num(muons)==2)
selections.add("leading_muon_pt", ak.firsts(muons.pt) > 30)
selections.add("subleading_muon_pt", ak.pad_none(muons, target=2)[:, 1].pt > 15)
selections.add("one_dimuon", ak.num(dimuons)==1)
region_mask = selections.all("golden_json", "npvsGood", "IsoMu24", "two_muons",
                            "leading_muon_pt", "subleading_muon_pt", "one_dimuon")
pruned_ev = events[region_mask]

muons = pruned_ev.Muon

good_muons = (
    (muons.pt > 10)
    & (abs(muons.eta) < 2.4)
    & muons.tightId
    & (muons.pfRelIso03_all < 0.15)
)
muons = muons[good_muons]

lorentz_muons = ak.zip(
    {
        "pt": muons.pt,
        "eta": muons.eta,
        "phi": muons.phi,
        "mass": muons.mass,
        "charge": muons.charge,
    },
    with_name="PtEtaPhiMCandidate",
    behavior=candidate.behavior,
)
dimuons = ak.combinations(lorentz_muons, 2, fields=["lead", "trail"])
dimuons = dimuons[ak.argsort(dimuons.lead.pt, axis=1)]
dimuons["p4"] = dimuons.lead + dimuons.trail
dimuons["pt"] = dimuons.p4.pt

good_dimuons = (
    (dimuons.lead.delta_r(dimuons.trail) > 0.3)
    & (dimuons.lead.charge * dimuons.trail.charge < 0)
    & (dimuons.p4.mass > 60.0) & (dimuons.p4.mass < 120.0)
)
dimuons = dimuons[good_dimuons]

################## event correction
cset_path = "constructors/corrections/sets/muon_Z.json.gz"
cname = "NUM_IsoMu24_DEN_CutBasedIdTight_and_PFIsoTight"
cset = correctionlib.CorrectionSet.from_file(cset_path)
corr = cset[cname]
m, n = ak.flatten(muons), ak.num(muons)
objects_pt_mask = m.pt > 26.01
objects_eta_mask = np.abs(m.eta) < 2.39
objects_mask = objects_pt_mask & objects_eta_mask
in_limit_objects = m.mask[objects_mask]
objects_pt = ak.fill_none(in_limit_objects.pt, 26.01)
objects_eta = ak.fill_none(np.abs(in_limit_objects.eta), 0.)
sf = corr.evaluate(objects_eta, objects_pt, "nominal")
sf = ak.where(objects_mask, sf, ak.ones_like(sf))
sf = ak.unflatten(sf, n)
sf = ak.prod(sf, axis=1)
weights_container = Weights(size=len(pruned_ev), storeIndividual=True)
weights_container.add("sf", sf)
region_weights = weights_container.weight()

#########filling and normalize histograms
weights = ak.broadcast_arrays(region_weights, dimuons.p4.mass)[0]
sumw_after = ak.sum(region_weights)
dimuon_mass_hist = hist.Hist.new.Reg(30, 60, 120, name="mass", label="mμμ [GeV]").Weight()
dimuon_mass_hist.fill(
    mass=ak.to_numpy(ak.flatten(dimuons.p4.mass)),
    weight=ak.to_numpy(ak.flatten(weights)),
)

L = 7980.4
xs = 6688
if is_mc:
    dimuon_mass_hist = dimuon_mass_hist * L * xs * sumw_after/sumw_before

out = {
    "mass": dimuon_mass_hist,
}
out["mass"].plot1d();
```
